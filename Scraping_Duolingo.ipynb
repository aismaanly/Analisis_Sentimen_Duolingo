```python
!pip install google-play-scraper

from google_play_scraper import app, reviews_all, Sort
import csv
import pandas as pd
```

```python


# Scraping review aplikasi Duolingo dari Play Store
scrapreview = reviews_all(
    'com.duolingo',
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=10000
)

# Mengubah hasil scraping ke DataFrame dan menyimpannya ke file CSV
df = pd.DataFrame(scrapreview)
df.to_csv('duolingo_review.csv', index=False)
```
